```python
from sklearn.metrics import silhouette_score, davies_bouldin_score, calinski_harabasz_score

CONFIG = {"match": ("ieeg", "match"), "avg_match": ("ieeg", "avg_match"),"avg_source": ("ieeg", "avg_source"),"source": ("ieeg", "source")}
LABEL_FILES = {"ieeg": "out/labels/labels_Kmean_singlePC3.csv",
                "match": "out/labels/labels_Kmean_singlePC3.csv",
                "avg_match": "out/labels/labels_Kmean_MEG_match_avg_pc.csv",
                "avg_source": "out/labels/labels_Kmean_MEG_source_avg_pc.csv",
                "source": "out/labels/labels_Kmean_MEG_source_pc.csv",}

def calculate_metrics(data, labels, dataset, n_components=3):
    res = []

    for pc in range(n_components) :
        if dataset == 'ieeg' : 
            label_col = 'ieeg'
        else :
            label_col = 'meg'

        res.append({"dataset": dataset,
                "pc_lab": f"pc{pc}",
                "ss": silhouette_score(data, labels[f"{label_col}_pc{pc}"]),
                "db": davies_bouldin_score(data, labels[f"{label_col}_pc{pc}"]),
                "ch": calinski_harabasz_score(data, labels[f"{label_col}_pc{pc}"])})
    return res

results = []
for model_dir, (x_dataset, y_dataset) in CONFIG.items():
    file = f"out/cross_models/cross_models_scale_{scaled_}_{model_dir}/{model}_model.p"
    with open(file, "rb") as f:
        cca_model = pickle.load(f)
    for data, dataset in zip((cca_model.x_weights_, cca_model.y_weights_), (x_dataset, y_dataset)):
        df_lab =pd.read_csv(LABEL_FILES[dataset]).drop(columns="Unnamed: 0")
        results.extend(calculate_metrics(data,df_lab, dataset))

metric_lab = pd.DataFrame(results)
metric_lab['model'] = ['match']*6 + ['avg_match']*6 + ['avg_source']*6 +  ['source']*6 
```
